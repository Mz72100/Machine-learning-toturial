# **Part 3: Preprocessing**

```python
from sklearn.preprocessing import LabelEncoder

# 1. Redefine categorical columns based on what's actually in the reduced dataframe
current_categorical_cols = train_reduced.select_dtypes(include=['object']).columns.tolist()

# 2. Safely encode only the columns that exist
train_encoded = train_reduced.copy()
for col in current_categorical_cols:
    if col in train_encoded.columns:
        le = LabelEncoder()
        train_encoded[col] = le.fit_transform(train_encoded[col].astype(str))

# 3. Define X and y by checking available columns
# We drop 'isFraud' and 'TransactionID', plus any other non-feature columns
cols_to_exclude = ['isFraud', 'TransactionID']
feature_cols = [c for c in train_encoded.columns if c not in cols_to_exclude]

X = train_encoded[feature_cols]
y = train_encoded['isFraud']

print(f"Data prepared. Total features: {len(feature_cols)}")
print(f"Target distribution: {y.value_counts(normalize=True)[1]:.2%} Fraud")
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import average_precision_score
import time
import numpy as np

# Using 3 folds for faster execution in Colab; 5 is standard for research
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

def evaluate_model(model, name, X_data, y_data):
    auprc_scores = []
    print(f"--- Training {name} ---")
    start_time = time.time()

    for fold, (train_idx, val_idx) in enumerate(skf.split(X_data, y_data)):
        X_train, X_val = X_data.iloc[train_idx], X_data.iloc[val_idx]
        y_train, y_val = y_data.iloc[train_idx], y_data.iloc[val_idx]

        # Fit model
        model.fit(X_train, y_train)

        # Predict probabilities for the Fraud class (1)
        preds = model.predict_proba(X_val)[:, 1]

        # Calculate AUPRC
        score = average_precision_score(y_val, preds)
        auprc_scores.append(score)
        print(f"Fold {fold+1} AUPRC: {score:.4f}")

    avg_score = np.mean(auprc_scores)
    runtime = time.time() - start_time
    print(f"Finished {name}. Avg AUPRC: {avg_score:.4f} | Time: {runtime:.2f}s\n")
    return avg_score, runtime
```

# **Part 5: Hyperparameter Optimization with HalvingGrid Search**

```python
from sklearn.experimental import enable_halving_search_cv
from sklearn.model_selection import HalvingGridSearchCV
import time

# Dictionary to store the best models and their stats
best_tuned_models = {}
tuning_results = []

def run_fast_tuning(name, estimator, param_grid):
    print(f"--- Starting Fast Tuning for {name} ---")
    start_time = time.time()

    # Using HalvingGridSearchCV for speed
    # factor=3 means we keep the top 1/3 performers in each round
    hgs = HalvingGridSearchCV(
        estimator=estimator,
        param_grid=param_grid,
        factor=3,
        resource='n_samples',
        max_resources='auto',
        scoring='average_precision', # Focusing on AUPRC for fraud
        cv=3,
        n_jobs=-1,
        random_state=42
    )

    hgs.fit(X_train, y_train)
    duration = time.time() - start_time

    print(f"{name} Tuning Complete in {duration:.2f}s")
    print(f"Best Params: {hgs.best_params_}\n")

    return hgs.best_estimator_, hgs.best_params_, duration
```

```python
# XGBoost Grid
xgb_grid = {
    'max_depth': [4, 6, 8],
    'learning_rate': [0.05, 0.1],
    'n_estimators': [100] # Keep low for tuning, increase for final fit
}

# LightGBM Grid
lgbm_grid = {
    'num_leaves': [31, 64],
    'learning_rate': [0.05, 0.1],
    'n_estimators': [100]
}

# CatBoost Grid
cat_grid = {
    'depth': [4, 6],
    'learning_rate': [0.05, 0.1],
    'iterations': [100]
}

# Define the models (using CPU as per your previous setting)
models_to_tune = {
    "XGBoost": (XGBClassifier(scale_pos_weight=25, tree_method='hist', random_state=42), xgb_grid),
    "LightGBM": (LGBMClassifier(is_unbalance=True, device='cpu', random_state=42, verbose=-1), lgbm_grid),
    "CatBoost": (CatBoostClassifier(auto_class_weights='Balanced', verbose=0, task_type="CPU", random_state=42), cat_grid)
}
```

# **Part 7: Class Balancing Technique**

## **Balancing With SMOTE**

```python
from imblearn.over_sampling import SMOTE
from collections import Counter
from sklearn.impute import SimpleImputer # Import SimpleImputer

# It is best to use a smaller ratio (e.g., sampling fraud to be 10-20% of data)
# rather than 50/50, to avoid creating too much noise.
sm = SMOTE(sampling_strategy=0.1, random_state=42)

print(f"Original training dataset shape: {Counter(y_train)}")

# Impute missing values before applying SMOTE
# Initialize the imputer - 'mean' is a common strategy for numerical data
imputer = SimpleImputer(strategy='mean')

# Fit on X_train and transform both X_train and X_val if you plan to use X_val later
# For SMOTE, we only need to transform X_train
X_train_imputed = imputer.fit_transform(X_train)

# Convert back to DataFrame, retaining column names if desired for clarity, though not strictly necessary for SMOTE
X_train_imputed_df = pd.DataFrame(X_train_imputed, columns=X_train.columns)

# Apply SMOTE to the imputed data
X_train_res, y_train_res = sm.fit_resample(X_train_imputed_df, y_train)

print(f"Resampled training dataset shape: {Counter(y_train_res)}")
```

```python
# Updated models without the 'unbalance' weights (since SMOTE does the balancing)
models_smote = {
    "XGBoost_SMOTE": XGBClassifier(n_estimators=300, tree_method='hist', random_state=42),
    "LightGBM_SMOTE": LGBMClassifier(n_estimators=300, device='cpu', random_state=42, verbose=-1),
    "CatBoost_SMOTE": CatBoostClassifier(iterations=300, verbose=0, task_type="CPU", random_state=42)
}

smote_performance = []

for name, model in models_smote.items():
    # Train on RESAMPLED data
    model.fit(X_train_res, y_train_res)

    # Evaluate on ORIGINAL VALIDATION data (Keep it realistic!)
    y_pred = model.predict(X_val)
    y_proba = model.predict_proba(X_val)[:, 1]

    smote_performance.append({
        "Model": name,
        "AUPRC": average_precision_score(y_val, y_proba),
        "Recall": recall_score(y_val, y_pred),
        "Precision": precision_score(y_val, y_pred),
        "F1-Score": f1_score(y_val, y_pred)
    })

# Compare results
smote_df = pd.DataFrame(smote_performance)
print(smote_df.to_string(index=False))
```

## **Balancing With Weight Ratio**

```python
# Calculate the exact ratio for the scale_pos_weight
count_legit = y_train.value_counts()[0]
count_fraud = y_train.value_counts()[1]
weight_ratio = count_legit / count_fraud

print(f"Calculated Weight Ratio: {weight_ratio:.2f}")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score

# Initialize models with their specific weighting arguments
models_weighted = {
    "XGBoost_Weighted": XGBClassifier(
        n_estimators=300,
        scale_pos_weight=weight_ratio, # Penalizes missing the positive class
        tree_method='hist',
        random_state=42
    ),

    "LightGBM_Weighted": LGBMClassifier(
        n_estimators=300,
        is_unbalance=True, # Automatically calculates weights based on data
        device='cpu',
        random_state=42,
        verbose=-1
    ),

    "CatBoost_Weighted": CatBoostClassifier(
        iterations=300,
        auto_class_weights='Balanced', # Balances weights automatically
        verbose=0,
        task_type="CPU",
        random_state=42
    )
}

weighted_performance = []

for name, model in models_weighted.items():
    # Train on ORIGINAL, UNMODIFIED data
    model.fit(X_train, y_train)

    y_pred = model.predict(X_val)
    y_proba = model.predict_proba(X_val)[:, 1]

    weighted_performance.append({
        "Model": name,
        "Accuracy": accuracy_score(y_val, y_pred),
        "AUC-ROC": roc_auc_score(y_val, y_proba),
        "AUPRC": average_precision_score(y_val, y_proba),
        "Recall": recall_score(y_val, y_pred),
        "Precision": precision_score(y_val, y_pred),
        "F1-Score": f1_score(y_val, y_pred)
    })

# Compare results
weighted_df = pd.DataFrame(weighted_performance)
print(weighted_df.to_string(index=False))
```
